```python
from logging import exception #hata türlerini ve loglarını yakalamak için kütüphane
import datetime # tarih ve saat bilgilerini aldığımız kütüphane
import pytz # İstanbul saatini almak için kullandığım kütüphane




gorevler= [] # ilk önce bu listeyi oluşturdum boş olmalı çünkü kullanıcıdan veri alacağım
tamamlanan_gorevler= [] # bunuda boş oluşturdum bu tamamlanmış görevleri tutacak



def dosya_kaydet(): # dosyaya kaydetme işlemleri için fonksiyon oluşturdum
  try:
    with open("gorevler.txt", "w", encoding="utf-8") as f: # görevler.txt dosyasını yazma modunda açtım çünkü listeden
    # görev silmek veya düzenlemek  istediğimde güncel listemiz gelsin diye
      for gorev in gorevler: # görevler listesinden  döngü ile görevleri  göreve   doldurur
        f.write(gorev + "\n") # f.write ile dosyaya görevleri yazar her görevin sonunda alt satıra geçer


    with open("tamamlanan_gorevler.txt", "w", encoding="utf-8") as f:
            for kayit in tamamlanan_gorevler: # tamamlanan_gorevler listesinden döngü ile tamamlanmış görevleri kayıdın içine atar
                f.write(kayit + "\n") # f.write ile dosyaya görevleri yazar


    print("💾 Dosyalar başarıyla kaydedildi.")
  except Exception as e: # hata türünü yakalar
    print(f"❌Dosyalar kaydedilemedi. Hata Türü: {type(e).__name__}") # hata mesajı ve hata türü ismini verir





def dosya_yukle():
  try:
    with open("gorevler.txt", "r", encoding="utf-8") as f: # dosyayı okuma modunda açıyorum uygulama başladığında listedeki görevleri okusun diye
      for satir in f: # dosyanın içindeki görevleri döngü ile satir a aktarır
        if satir.strip(): # .strip yazının önündeki arkasındaki gizli karakterleri temizler örneğin \n
          gorevler.append(satir.strip()) # temizlenmiş veri görevler listesine eklenir
          print("📂 'gorevler.txt' yüklendi.")
  except FileNotFoundError: # dosya var mı yok mu diye kontrol eder ya da herhangi başka bir hata durumu var mı ona bakar
    print("ℹ️ Henüz kayıtlı görev yok, yeni liste ile başlanıyor.") # hata mesajı


  try:
    with open("tamamlanan_gorevler.txt", "r", encoding="utf-8") as f:
      for satir in f:
        if satir.strip():
          tamamlanan_gorevler.append(satir.strip()) # daha önceden tamamlanan görevleri dosyadan alır tamamlanan_gorevler listesine ekler
  except FileNotFoundError:
    pass





def gorev_sil():
  print("\n --- 🗑️Görev Silme ---")
  gorev_listele() # listeden seçim yapılacağı için ilk önce listele fonksiyonunu çağırıyoruz
  if len(gorevler) == 0: # gorevlerin uzunluğunu kontrol ediyor boşsa geri döndürüyor
    return
  try: # hatalı girişlerin kontrolü için
    secim = int(input("Silmek istediğiniz görevin numarasını girin: ")) # listeden numara ile seçim yapılacağı için int input kullandım
    if 1 <= secim <= len(gorevler):# seçimin görevler listesindeki numaralar arasında olup olmadığının kontrolü
      silinen_gorev = gorevler.pop(secim-1) # .pop() fonksiyonu ile verilen inidsteki eleman silinir ve silinen elemanı gösterir
      dosya_kaydet() # değişiklik anında dosyaya etki eder
      print(f"{silinen_gorev} görevi başarıyla silindi. 🗑️")
    else:
      print("❌Hata: Böyle bir görev numarası yok.") # listede olmayan bir görev numarası girilirse kullanıcıya hata mesajı verir
  except Exception as c : #hata türünü yakalar
    print(f"❌Hata: Lütfen sayı giriniz. Hata Türü: {type(c).__name__}") # hata mesajı ve hata türü ismini kullanıcıya gösterir





def gorev_duzenle():
    print("\n --- 📝Görev Düzenleme ---")
    gorev_listele()
    if len(gorevler) == 0:
        return

    try:
        secim = int(input("Düzenlenecek görevin numarası: "))
        if 1 <= secim <= len(gorevler):
            eski_hal = gorevler[secim - 1]
            print(f"Eski hali: {eski_hal}") # değiştirilecek görevin eski adı kullanıcıya gösterilir

            yeni_hal = input("📝Yeni hali ne olsun?: ")
            if yeni_hal.strip() != "": # yeni hali boş değilse devam eder
                gorevler[secim - 1] = yeni_hal # listedeki ilgili sıradaki görevi günceller
                dosya_kaydet() # değişiklik anında dosyaya etki eder
                print("Görev güncellendi.✅")
            else:
                print("❌Hata: 🫗Boş bırakamazsınız.") # boş bir değer girildiyse hata mesajı verir
        else:
            print("❌Hata: Geçersiz numara.") # listede olmayan bir görev numarası girilirse kullanıcıya hata mesajı verir
    except Exception as b:
        print(f"❌Hata: Lütfen sayı girin. Hata Türü: {type(b).__name__}")




def gorev_ekle():
  print("\n --- 📩Görev Ekleme ---")
  yeni_gorev= input("Yapılacak görevi giriniz: ")
  if yeni_gorev.strip() == "": # boş değilse devam eder
        print("❌Hata: 🫗 Boş bir görev ekleyemezsiniz!")  # boş bir değer girildiyse hata mesajı verir
  else:
        gorevler.append(yeni_gorev)
        dosya_kaydet() # değişiklik anında dosyaya etki eder
        print(f"{yeni_gorev} görevi başarıyla eklendi. 📩")





def gorev_listele():
  print("\n --- 📑Görev Listesi ---")
  if len(gorevler) == 0:
    print("🫗Görev listesi boş.")
  else:
    for numara, gorev in enumerate(gorevler,1): # enumerate fonksiyonu hem sıra numarasını hem görevi verir. normalde listeler indis ile başlar yani 0 dan enumarate 1 den başlar
      print(f"{numara}. {gorev}")





def gorev_tamamla():
    print("\n --- 📋Görev Tamamlama ---")
    gorev_listele()

    if len(gorevler) == 0:
        return

    try:
        secim = int(input("Tamamlanan görevin numarası: "))

        if 1 <= secim <= len(gorevler):
            biten_gorev = gorevler.pop(secim - 1) # görev görev listesinden silinir ve .pop ile biten_gorev e aktarılır
            istanbul_tz = pytz.timezone('Europe/Istanbul') # İstanbul saat dilimini tanımlıyorum
            simdi = datetime.datetime.now(istanbul_tz)
            zaman_damgasi = simdi.strftime("%d.%m.%Y %H:%M") # Tarihi okunabilir formata (Gün.Ay.Yıl Saat:Dakika) çeviriyorum
            kayit = f"{biten_gorev} (Tamamlandı: {zaman_damgasi})" # Görev isminin yanına tarihi ekleyerek yeni bir kayıt metni oluşturuyorum ve bunu kayit değişkenine atıyorum
            tamamlanan_gorevler.append(kayit) # tamamlanan_gorevler listesine kayit nesnesindekileri ekliyorum
            dosya_kaydet()  # değişiklik anında dosyaya etki eder tamamlanan görev hem gorevler listesinden hem de gorevler.txt den silinir ve tamamlanan_gorevler.txt ye yazılır

            print(f"✅Tebrikler! '{biten_gorev}' tamamlandı.")
        else:
            print("❌Hata: Böyle bir görev numarası yok.")

    except Exception as a:
        print(f"❌Hata: Lütfen sadece sayı girin. Hata Türü: {type(a).__name__}") #farklı bir hata türü ile karşılaşırsak diye hata türü adı ekledim




def tamamlanan_gorevleri_goster():
    print("\n --- 📜Tamamlanan Görevler ---")
    if len(tamamlanan_gorevler) == 0:
        print("Henüz tamamlanan bir görev yok.")
    else:
        for numara, gorev in enumerate(tamamlanan_gorevler, 1):
            print(f"{numara}. {gorev}")





def menu_goster():
  dosya_yukle() # program başalr başlamaz dosyadan verileri yükler

  while True:
   print("\n --- TO DO LİST UYGULAMASINA HOŞ GELDİNİZ  ---")
   print("1- 📑Görevleri Göster")
   print("2- 📩Görev Ekle")
   print("3- 📝Görevleri Düzenle")
   print("4- 🗑️Görevleri Sil")
   print("5- 🚪Çıkış")
   print("6- 📋Tamamlandı")
   print("7- 📜Tamamlanan Görevleri Göster")



   try:

    secim = int (input("Yapmak istediğiniz işlemin numarası (1-7) : "))
    if secim == 1:
     gorev_listele()
    elif secim == 2:
        gorev_ekle()
    elif secim == 3:
        gorev_duzenle()
    elif secim == 4:
        gorev_sil()
    elif secim == 5:
        print("🚪Çıkış yapılıyor...")
        break
    elif secim == 6:
        gorev_tamamla()
    elif secim == 7:
        tamamlanan_gorevleri_goster()
    else:
        print("❌Hata: Hatalı seçim! Lütfen 1-7 arası bir sayı girin.")
   except Exception as e:
    print(f"❌Hata: Lütfen sayı giriniz. Hata Türü: {type(e).__name__}")

menu_goster()


```
